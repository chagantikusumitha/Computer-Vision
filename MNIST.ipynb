```python
import os
from PIL import Image
import numpy as np
import pandas as pd
from os import listdir
 

directory = "/content/mnist1"
l = []
l1 = []
for item in os.listdir(directory):
  folder_dir = os.path.join(directory, item)
# get the path/directory
  
  for i in os.listdir(folder_dir):
       img = Image.open(os.path.join(folder_dir, i))
       l.append(list(np.array(img).flatten()))
       l1.append(i[0])
df = pd.DataFrame(l)
df["label"] = l1
df.to_csv("pil1")
```

```python
df = pd.read_csv("/content/drive/MyDrive/cv1/cv")
```

```python
df.label.nunique()
```

```python
X = df.drop('label', axis=1)

y = df['label']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=100)
print(X_train.shape, y_train.shape)
print(X_test.shape, y_test.shape)
```

Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
classifier = LogisticRegression()
classifier.fit(X_train, y_train)
```

```python
y_test_pred = classifier.predict(X_test)
```

```python
from sklearn import metrics

metrics.accuracy_score(y_test, y_test_pred)
```

Decision Tree Classifier

```python
from sklearn import tree
classifier = tree.DecisionTreeClassifier()
classifier.fit(X_train, y_train)
```

```python
y_test_pred = classifier.predict(X_test)
```

```python
from sklearn import metrics

metrics.accuracy_score(y_test, y_test_pred)
```

KNN Classification

```python
from sklearn.neighbors import KNeighborsClassifier
classifier = KNeighborsClassifier()
classifier.fit(X_train, y_train)
```

```python
y_test_pred = classifier.predict(X_test)
from sklearn import metrics
metrics.accuracy_score(y_test, y_test_pred)
```

Support Vector Classifier

```python
from sklearn.svm import SVC
classifier = SVC()
classifier.fit(X_train, y_train)
```

```python
y_test_pred = classifier.predict(X_test)
from sklearn import metrics
metrics.accuracy_score(y_test, y_test_pred)
```

Naive Bayes Classifier

```python
from sklearn.naive_bayes import GaussianNB
classifier = GaussianNB()
classifier.fit(X_train, y_train)
```

```python
y_test_pred = classifier.predict(X_test)
from sklearn import metrics
metrics.accuracy_score(y_test, y_test_pred)
```

Random Forest Classifier

```python
from sklearn.ensemble import RandomForestClassifier
classifier = RandomForestClassifier()
classifier.fit(X_train, y_train)
```

```python
y_test_pred = classifier.predict(X_test)
from sklearn import metrics
metrics.accuracy_score(y_test, y_test_pred)
```

Gradient Boosting Classifier

```python
from sklearn.ensemble import GradientBoostingClassifier
classifier = GradientBoostingClassifier()
classifier.fit(X_train, y_train)
```

```python
y_test_pred = classifier.predict(X_test)
from sklearn import metrics
metrics.accuracy_score(y_test, y_test_pred)
```

Ada Boost Classifier

```python
from sklearn.ensemble import AdaBoostClassifier
classifier = AdaBoostClassifier()
classifier.fit(X_train, y_train)
```

```python
y_test_pred = classifier.predict(X_test)
from sklearn import metrics
metrics.accuracy_score(y_test, y_test_pred)
```
